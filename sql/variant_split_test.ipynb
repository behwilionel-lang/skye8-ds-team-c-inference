⦁	For each experiment, test whether the observed variant split matches the planned split. State the test, the statistic and the p-value. This is a diagnostic, not a hypothesis about the product.


This is done with the use of the chi-square goodness-of-fit test

```python
import pandas as pd
from scipy.stats import chisquare
```

```python
df_vst = pd.read_csv("variant_split_test.csv")
df_vst
```

```python
results = []
for exp_id, group in df_vst.groupby('experiment_id'):
  observed = group['observed_count'].tolist()
  total = sum(observed)
  expected = [total * 0.5, total * 0.5]
  stat, p = chisquare(f_obs=observed, f_exp=expected)
  results.append({
    'experiment_id': exp_id,
    'observed_control': group[group['variant'] == 'control']['observed_count'].values[0],
    'observed_treatment': group[group['variant'] == 'treatment']['observed_count'].values[0],
    'chi2_stat': round(stat, 4),
    'p_value': round(p, 4),
    'verdict': 'FAIL' if p<0.05 else 'PASS' 
  })

df_results = pd.DataFrame(results)
print(df_results)
```
